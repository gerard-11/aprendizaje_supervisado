# 3. Feature Engineering
Convertimos categorías y normalizamos variables numéricas.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, QuantileTransformer

df = pd.read_csv('./data/titanic_clean.csv')
features = df.drop(columns=['Survived']).copy()
target = df['Survived']
features['Sex'] = LabelEncoder().fit_transform(features['Sex'])
features['Embarked'] = LabelEncoder().fit_transform(features['Embarked'])
transformer = QuantileTransformer(output_distribution='normal', random_state=100)
features[['Age', 'Fare']] = transformer.fit_transform(features[['Age', 'Fare']])
engineered = pd.concat([features, target], axis=1)
engineered.to_csv('./data/titanic_features.csv', index=False)
engineered.head()
```
