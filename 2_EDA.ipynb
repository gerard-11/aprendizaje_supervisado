# 2. Análisis exploratorio de datos
Exploramos la distribución de las variables y su relación con la supervivencia.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('./data/titanic_clean.csv')
display(df.describe(include='all'))
print(df.isna().sum())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.countplot(data=df, x='Survived', ax=axes[0])
sns.barplot(data=df, x='Sex', y='Survived', ax=axes[1])
sns.barplot(data=df, x='Pclass', y='Survived', ax=axes[2])
plt.tight_layout()
```

```python
sns.histplot(data=df, x='Age', hue='Survived', kde=True, bins=30, element='step')
plt.title('Distribución de edad por supervivencia')
plt.show()
```
